```python
#1.Min-Max Search Algorithm for Chess Game:

def minimax(depth, node, is_max):
    if depth == 2:
        return scores[node]

    if is_max:
        best = -999
        for i in range(2):
            value = minimax(depth + 1, node * 2 + i, False)
            best = max(best, value)
        return best

    else:
        best = 999
        for i in range(2):
            value = minimax(depth + 1, node * 2 + i, True)
            best = min(best, value)
        return best


scores = [3, 5, 2, 9]

result = minimax(0, 0, True)

print("Best score for Chess:", result)
```

```python
#2.Alpha-Beta Pruning Algorithm for Chess Game:

def alphabeta(depth, node, alpha, beta, is_max):

    if depth == 2:
        return scores[node]

    if is_max:
        best = -999

        for i in range(2):
            value = alphabeta(depth + 1, node * 2 + i,
                              alpha, beta, False)

            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:
        best = 999

        for i in range(2):
            value = alphabeta(depth + 1, node * 2 + i,
                              alpha, beta, True)

            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                break

        return best


scores = [3, 5, 2, 9]

result = alphabeta(0, 0, -999, 999, True)

print("Best score for Chess:", result)
```

```python
#3.Min-Max Search Algorithm for Tic-Tac-Toe Game:

def print_board():
    for i in range(0, 9, 3):
        print(board[i], "|", board[i+1], "|", board[i+2])
        if i < 6:
            print("---------")


def check_winner():
    win = [
        [0,1,2], [3,4,5], [6,7,8],
        [0,3,6], [1,4,7], [2,5,8],
        [0,4,8], [2,4,6]
    ]

    for a, b, c in win:
        if board[a] == board[b] == board[c] != " ":
            return board[a]

    return None


def minimax(is_max):
    winner = check_winner()

    if winner == "X":
        return 1
    if winner == "O":
        return -1
    if " " not in board:
        return 0

    if is_max:
        best = -999

        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                score = minimax(False)
                board[i] = " "
                best = max(best, score)

        return best

    else:
        best = 999

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(True)
                board[i] = " "
                best = min(best, score)

        return best


def best_move():
    best_score = -999
    move = -1

    for i in range(9):
        if board[i] == " ":
            board[i] = "X"
            score = minimax(False)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move


board = [" "] * 9

print("Tic-Tac-Toe Game")
print("You are O and AI is X")

while True:
    print_board()

    pos = int(input("Enter position (1-9): ")) - 1

    if pos < 0 or pos > 8 or board[pos] != " ":
        print("Invalid position! Try again.")
        continue

    board[pos] = "O"

    if check_winner() == "O":
        print_board()
        print("You Win!")
        break

    if " " not in board:
        print_board()
        print("Game Draw!")
        break

    move = best_move()
    board[move] = "X"

    if check_winner() == "X":
        print_board()
        print("AI Wins!")
        break

    if " " not in board:
        print_board()
        print("Game Draw!")
        break
```
